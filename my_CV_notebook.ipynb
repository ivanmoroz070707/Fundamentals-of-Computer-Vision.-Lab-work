```python
import cv2 #opencv
import numpy as np # matrix manipulations

%matplotlib inline 
from matplotlib import pyplot as plt 
```

## Image load

```python
input_image = cv2.imread('1637824248_4-koshka-top-p-multyashnii-kotik-4.jpg')
b,g,r = cv2.split(input_image)
```

## Image show channels

```python
plt.imshow(r, cmap='gray')
```

## Image color

```python
merged_img = cv2.merge([r,g,b])
plt.imshow(merged_img)
```

## Image crop

```python
crop_image = merged_img[0:300, 300:580, :]
plt.imshow(crop_image)
```
